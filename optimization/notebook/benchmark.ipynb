```python
import sys
sys.path.append('../../')
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
from core.evaluator import lvdt_evaluator, vc_evaluator
```

```python
lvdt_json_filename = "../config/initial_config/v0_typeG_lvdt.json"
vc_json_filename = "../config/initial_config/v0_typeG_vc.json"

lvdt_opt_json_filename = "../config/benchmark/opt_typeG_lvdt.json"
vc_opt_json_filename = "../config/benchmark/opt_typeG_vc.json"

lvdt_output_dir = "../data/benchmark/"
vc_output_dir = "../data/benchmark/"

lvdt_output_filename = "opt_typeG_LVDT.h5"
vc_output_filename = "opt_typeG_VC_.h5"
```

```python

# params: [0: NLI, 1: DI,2: LI,3: NLO,4: DO,5: LO,6: DIO]
params = [6, 14, 20, 7, 34, 25, 17.5]
f1, f2 = lvdt_evaluator(
    params = params, 
    input_json_filename = lvdt_json_filename,
    iter_json_filename=lvdt_opt_json_filename,
    output_dir = lvdt_output_dir,
    output_filename = lvdt_output_filename
)
    
f3, f4 = vc_evaluator(
    params = params,
    input_json_filename = vc_json_filename,
    iter_json_filename=vc_opt_json_filename,
    output_dir = vc_output_dir,
    output_filename = vc_output_filename
)

print(f1, f2, f3, f4)
```

```python

with open("../solutions/benchmark.txt", "w") as f:
    f.write(f"{'S':>10} {'NLE':>10} {'MF':>10} {'FVI':>10}\n")
    f.write(f"{f1:>10.4f} {f2:>10.4f} {f3:>10.4f} {f4:>10.4f}\n")
```
