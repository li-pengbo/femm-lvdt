#### pymoo

```python
import numpy as np
from pymoo.core.problem import ElementwiseProblem   
class MyProblem(ElementwiseProblem):

    def __init__(self):

        x_min = np.zeros(7)
        x_max = np.zeros(7)

        super().__init__(n_var=7,n_obj = 4, n_constr=0, xl=x_min, xu=x_max)
        
    def _evaluate(self, x, out, *args, **kwargs):
        f1, f2 = lvdt_evaluator(x, lvdt_json_filename, lvdt_iter_filename, lvdt_output_dir, lvdt_output_filename)
        f3, f4 = vc_evaluator(x, vc_json_filename, vc_iter_filename, vc_output_dir, vc_output_filename)

        out["F"] = [f1,f2, f3, f4]
```
